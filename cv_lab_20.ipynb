```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read image in grayscale
img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

# Define the structuring element (kernel) for dilation
# A larger kernel size will result in more pronounced dilation
kernel_size = 5
kernel = np.ones((kernel_size, kernel_size), np.uint8)

# Perform Dilation
dilated_img = cv2.dilate(img, kernel, iterations=1)

# Display Original vs Dilated Image
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(dilated_img, cmap='gray')
plt.title(f"Dilated Image (Kernel: {kernel_size}x{kernel_size})")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files
import os

# 1. Use an uploaded video file or ask the user to upload one
video_path = 'video.mp4'
if not os.path.exists(video_path):
    print("Please upload a video file (.mp4, .avi, etc.):")
    uploaded = files.upload()
    if uploaded:
        video_path = list(uploaded.keys())[0]
    else:
        video_path = None

if video_path and os.path.exists(video_path):
    cap = cv2.VideoCapture(video_path)
    frames = []
    max_frames_to_capture = 4

    # Read a few frames to display inline as an alternative to interactive cv2.imshow
    while cap.isOpened() and len(frames) < max_frames_to_capture:
        ret, frame = cap.read()
        if not ret:
            break
        # Convert BGR to RGB
        frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        frames.append(frame_rgb)

    cap.release()

    # Render the captured video frames side-by-side
    if frames:
        plt.figure(figsize=(15, 5))
        for idx, f in enumerate(frames):
            plt.subplot(1, len(frames), idx + 1)
            plt.imshow(f)
            plt.title(f"Frame {idx + 1}")
            plt.axis("off")
        plt.show()
    else:
        print("Could not read any frames from the video file.")
else:
    print("No video file is available to process.")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
import os

# 1. Use an already uploaded image or ask the user to upload one
path = 'image.jpg'
if not os.path.exists(path):
    print("Please upload an image:")
    uploaded = files.upload()
    if uploaded:
        path = list(uploaded.keys())[0]
    else:
        path = None

if path and os.path.exists(path):
    # Read the image and convert from BGR to RGB
    img = cv2.imread(path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # 2. Set up source and destination points for Affine Transform
    h, w = img_rgb.shape[:2]
    a = np.float32([[0, 0], [w - 1, 0], [0, h - 1]])
    b = np.float32([[50, 50], [w - 50, 50], [50, h - 50]])

    # Calculate matrix and perform warp
    m = cv2.getAffineTransform(a, b)
    out = cv2.warpAffine(img_rgb, m, (w, h))

    # 3. Use Matplotlib to render inline instead of cv2.imshow
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))

    axes[0].imshow(img_rgb)
    axes[0].set_title("Original Image")
    axes[0].axis("off")

    axes[1].imshow(out)
    axes[1].set_title("Affine Transformed")
    axes[1].axis("off")

    plt.tight_layout()
    plt.show()
else:
    print("No image file is available to process.")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# Use 'image.jpg' which is already available in your environment
path = 'image.jpg'

if os.path.exists(path):
    img = cv2.imread(path)
    # Convert BGR to RGB for correct Matplotlib rendering
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    h, w = img_rgb.shape[:2]
    # Define the 4 corner points for Perspective Transform
    a = np.float32([[0, 0], [w - 1, 0], [0, h - 1], [w - 1, h - 1]])
    b = np.float32([[50, 50], [w - 50, 50], [50, h - 50], [w - 50, h - 50]])

    # Calculate perspective transform matrix and warp
    m = cv2.getPerspectiveTransform(a, b)
    out = cv2.warpPerspective(img_rgb, m, (w, h))

    # Render the input and perspective warp inline using Matplotlib
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))

    axes[0].imshow(img_rgb)
    axes[0].set_title("Original Image")
    axes[0].axis("off")

    axes[1].imshow(out)
    axes[1].set_title("Perspective Warp")
    axes[1].axis("off")

    plt.tight_layout()
    plt.show()
else:
    print("image.jpg was not found. Please upload an image first.")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# Use 'image.jpg' which is already available in your environment
path = 'image.jpg'

if os.path.exists(path):
    img = cv2.imread(path)
    # Convert BGR to RGB for correct colors in Matplotlib
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Convert to grayscale for Harris Corner detection
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Harris Corner Detection
    dst = cv2.cornerHarris(gray, 2, 3, 0.04)

    # Dilate corner results to make them larger/clearer in visualization
    dst = cv2.dilate(dst, None)

    # Threshold for marking the corners in red on the RGB image
    img_rgb[dst > 0.01 * dst.max()] = [255, 0, 0]

    # Render the result inline using Matplotlib
    plt.figure(figsize=(8, 8))
    plt.imshow(img_rgb)
    plt.title("Harris Corners (Red)")
    plt.axis("off")
    plt.show()
else:
    print("image.jpg was not found. Please run an earlier cell to upload it.")
```
